```python
#adding legend
x = [1,2,3,4,5]
y = [1,4,9,16,25]
plt.plot(x,y, marker = "x", linestyle = "--")
plt.title("Basic line plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend(["y=x^2"])
plt.show()
```

```python
#adding grid line
x = [1,2,3,4,5]
y = [1,4,9,16,25]
plt.plot(x,y, marker = "x", linestyle = "--")
plt.title("Basic line plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend(["y=x^2"])
plt.grid()
plt.show()
```

```python
#adjusting limits of axis
x = [1,2,3,4,5]
y = [1,4,9,16,25]
plt.plot(x,y, marker = "x", linestyle = "--")
plt.title("Basic line plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(4,6)
plt.ylim(15,40)
plt.legend(["y=x^2"])
plt.grid()
plt.show()
```

```python
#Line plot with custom sticks
x = [1,2,3,4]
y = [50,150,200,120]
plt.plot(x,y)
plt.xticks(ticks = [1,2,3,4], labels = ["June","July","August","September"])
plt.yticks(ticks = y, labels = ["Low","Medium","High","V.High"])
plt.show()
```

```python
#adding multiple lines
x = [1,2,3,4,5]
y = [1,3,5,7,9]
y2 = [2,4,6,8,10]
plt.plot(x,y)
plt.plot(x,y2, color = "Blue", linestyle = ":")
plt.title("Multiple Plots")
plt.legend(["Plot 1","Plot 2"])
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
y1 = [1,3,5,7,9]
y2 = [3,2,5,9,8]
y3 = [6,5,8,7,9]

plt.subplot(2,2,1)
plt.plot(x,y, color = "red", label = "First")
plt.title("Plot 1")
plt.legend()

plt.subplot(2,2,2)
plt.plot(x,y1, color = "green", label = "Second")
plt.title("Plot 2")
plt.legend()

plt.subplot(2,2,3)
plt.plot(x,y2, color = "purple", label = "Third")
plt.title("Plot 3")
plt.legend()

plt.subplot(2,2,4)
plt.plot(x,y3, color = "blue", label = "Fourth")
plt.title("Plot 4")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
#with scatter
x = [1,2,3,4,5]
y = [2,4,6,8,10]
y1 = [1,3,5,7,9]
y2 = [3,2,5,9,8]
y3 = [6,5,8,7,9]

plt.subplot(2,2,1)
plt.plot(x,y, color = "red", label = "First")
plt.title("Plot 1")
plt.legend()

plt.subplot(2,2,2)
plt.scatter(x,y1, color = "green", label = "Second")
plt.title("Plot 2")
plt.legend()

plt.subplot(2,2,3)
plt.scatter(x,y2, color = "purple", label = "Third")
plt.title("Plot 3")
plt.legend()

plt.subplot(2,2,4)
plt.plot(x,y3, color = "blue", label = "Fourth")
plt.title("Plot 4")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
fig,axes=plt.subplots(2,1)
print(axes)
# First Figure
axes[0].plot(x,y,color="Orange",label="first")
axes[0].set_title("Top plot")
axes[0].set_xlabel("X1")
axes[0].set_ylabel("Y1")
axes[0].legend()
axes[0].grid()


#second Figure
axes[1].plot(x,y,color="blue",label="Second")
axes[1].set_title("Bottom plot")
axes[1].set_xlabel("X2")
axes[1].set_ylabel("Y2")
axes[1].legend()
axes[1].grid()
plt.tight_layout()
plt.show()
```
