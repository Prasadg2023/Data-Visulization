# Q.1) Convert the following data into the dataset 
* data = {'Age' : [24,35,29,30,26,45,53,65,71,13]}
    * Write the insights for the following plots. 
    * Plot the histogram on the above data 
    * Plot kde plot on the given data
    * Plot the box plot on the given data

```python
import pandas as pd
data = {'Age' : [24,35,29,30,26,45,53,65,71,13]}
df = pd.DataFrame(data)
df
```

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot  as plt
import seaborn as sns
import plotly.express as px
```

```python
# Plot the histogram on the above data 
sns.histplot(data)
plt.show()
```

# Insights from the Plots
# Histogram – Age Distribution

* Most ages are clustered between 20 and 40 years.

* here are fewer observations at very young (13) and very old ages (65, 71).

* The distribution is right-skewed, meaning higher age values stretch the tail to the right.

```python
# Plot kde plot on the given data
sns.kdeplot(data)
plt.show()
```

# KDE Plot – Density of Age
* The KDE curve peaks around 25–35 years, showing this is the most common age range.

* The smooth curve confirms the right-skewed nature of the data.

* Older ages form a long tail, indicating fewer but extreme values.

```python
# Plot the box plot on the given data
sns.boxplot(data)
plt.show()
```

# Box Plot – Spread & Outliers

* The median age lies around the early 30s.

* The interquartile range (IQR) shows most data is between young and middle-aged adults.

* Ages like 65 and 71 appear as upper outliers, indicating unusually high values.

* Age 13 lies on the lower end but within acceptable range.

      “Marks” : [90, 85, 80, 75, 86,65,72,89,76,84]}

```python
Data = {"names": ["jhon", "alex", "jeremy", "jenney", "sam", "joe", "ruth", "emily", "sophia", "michael"],
        "Marks" : [90, 85, 80, 75, 86,65,72,89,76,84]}
pd.DataFrame(Data)
```

```python
# Plot vertical and horizontal bar plot and write the insights from the plot, what you have observed.
plt.bar(Data["names"], Data["Marks"])
plt.show()
```

```python
# Horizontal
plt.barh(Data["names"], Data["Marks"])
plt.show()
```

# Insights (Vertical Bar Plot)
* Jhon scored the highest marks (90).
* Joe scored the lowest marks (65).
* Most students scored between 75 and 90.
* Marks distribution is fairly balanced.

# Insights (Horizontal Bar Plot)
* Easier to compare marks across students.
* Top performers (Jhon, Emily, Sam) are clearly visible.
* Helps identify lowest scorer (Joe) quickly.
* Suitable when category names are long.
