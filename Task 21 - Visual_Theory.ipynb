++
Advantage:

**Q2 Plots for Correlation Analysis**

**Identify the best plot for analyzing correlation between variables. Briefly explain this plot, its features, and key parameters that can be adjusted to enhance clarity and interpretability.**


**Best Plot for Correlation Analysis: Scatter Plot**

The scatter plot is the most effective and commonly used plot for analyzing the correlation between two numerical variables.

**Explanation of Scatter Plot**

A scatter plot displays individual data points on a 2D plane, where:

+ The x-axis represents one variable

+ The y-axis represents another variable

Each point corresponds to one observation, making it easy to see how the two variables move in relation to each other.

**Features of a Scatter Plot**

+ Shows direction of correlation:

     + Upward trend → Positive correlation

     + Downward trend → Negative correlation
 
     + No clear pattern → Weak or no correlation

+ Reveals strength of relationship (tight or loose clustering of points)

+ Identifies outliers easily

+ Supports trend lines for better interpretation

**Key Parameters to Enhance Clarity and Interpretability**

**1 Marker size (s)**

Adjust point size to reduce overcrowding

Example: smaller points for large datasets

**2 Color (c)**

Different colors can represent categories or intensity

Helps distinguish groups in the data

**3 Transparency (alpha)**

Useful when points overlap

Makes dense areas more visible

**4 Trend line / Regression line**

Adding a best-fit line highlights the correlation direction and strength

**5 Axis labels and title**

Clearly describe the variables being compared

**6 Grid (grid=True)**

Improves readability and value estimation

**Pie and Donut Charts in Data Visualization**

Pie and donut charts are widely used to represent proportions or percentages of a whole. They are most effective for categorical data where the goal is to show the relative size of each category.

**1. Types of Data Suited**

**Categorical data:** Data that represents distinct categories or groups.
Examples:

  + Sales by product category

  + Market share of companies

  + Votes per political party

**Percentage or proportion data:** Pie and donut charts work best when the sum of all categories represents 100% of a whole

2. Differences Between Pie and Donut Charts
| Feature             | Pie Chart                       | Donut Chart                                  |
| ------------------- | ------------------------------- | -------------------------------------------- |
| Appearance          | Full circle divided into slices | Circle with a hole in the center             |
| Use                 | Show proportions of a whole     | Same as pie chart, but more modern / stylish |
| Center Space        | No empty space                  | Empty center can display labels or metrics   |
| Customization Focus | Slice colors, labels, explode   | Same as pie + central text                   |




**3. Python Syntax**

**Pie Chart**
import matplotlib.pyplot as plt

# Example data
labels = ['A', 'B', 'C', 'D']

sizes = [25, 35, 20, 20]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, shadow=True)

plt.title("Pie Chart Example")

plt.show()

**Donut Chart**

# Create a donut chart

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, wedgeprops={'width':0.4})

plt.title("Donut Chart Example")

plt.show()

**4. Key Customizable Parameters**

1 labels – Names of categories.

2 sizes – Numerical values for each category.

3 autopct – Display percentage inside slices.

4startangle – Rotate the chart for better orientation.

5 shadow – Adds a shadow for 3D effect.

6 colors – Customize slice colors.

7 explode – Highlight a slice by “pulling” it out.

8 wedgeprops (for donut) – Adjust width, edge color, etc.

**Box Plots**
  
**Describe box plots, including their structure, types, and use cases in data analysis. Explain where box plots are typically used and explore their main parameters.**



A box plot (or box-and-whisker plot) is a graphical representation of a data distribution, highlighting median, quartiles, and outliers.
It is commonly used in exploratory data analysis (EDA) to quickly understand the spread and symmetry of data

**Structure of a Box Plot**

+ Minimum (lower whisker) – smallest non-outlier value

+ Q1 (1st quartile / 25%) – lower edge of the box

+ Median (Q2 / 50%) – line inside the box

+ Q3 (3rd quartile / 75%) – upper edge of the box

+ Maximum (upper whisker) – largest non-outlier value

+ Outliers – points outside 1.5×IQR (interquartile range)

**. Types of Box Plots**

Vertical Box Plot – default, box drawn along Y-axis

Horizontal Box Plot – box drawn along X-axis (orient='h')

Grouped / Comparative Box Plot – multiple boxes for different categories

**Use Cases**

Detect outliers

Understand data spread and skewness

Compare distributions between groups or categories

Useful in statistical summaries and EDA

```python
import matplotlib.pyplot as plt

data = [7, 8, 5, 6, 10, 12, 4, 6, 8]
plt.boxplot(data)
plt.title("Box Plot Example")
plt.show()
```
